**Emperical Rule**

 - 68% of data available between u-1*sd to u+1*sd

- step-1: calculate mean of wage: mean=74455

- step-2: calculate sd of wage: sd: 52815

- step-3: lb=mean-1*sd : 21640

- step-4: ub=mean+1*sd  : 127270

- step-5: con1= wage_data>lb

- step-6: con2= wage_data<ub

- step-7: con= con1 & con2

- step-8: np.sum(con): 17170

- step-9: 68%(25480) : 96328

```python
# step-1: calculate mean of wage: mean=74455
wage_data=visa_df['prevailing_wage']
mean=wage_data.mean()
#step-2: calculate sd of wage: sd: 52815
sd=wage_data.std()
#step-3: lb=mean-1*sd : 21640
lb=mean-1*sd
#step-4: ub=mean+1*sd  : 127270
ub=mean+1*sd
#step-5: con1= wage_data>lb
con1= wage_data>lb
#step-6: con2= wage_data<ub
con2= wage_data<ub
# step-7: con= con1 & con2
con= con1 & con2
# step-8: np.sum(con): 17170
np.sum(con)==68*(25480)/100
```
