```python
import numpy as np
import pandas as pd
```

```python
d={'Names':['Ramesh','Suresh',None,'Mahesh'],
  'Age':[None,22,23,25],
  'City':['hyd','blr','pune',None]}
df1=pd.DataFrame(d)
df1
```

```python
df1.to_csv('data1.csv',index=False)
```

```python
d={'Names':['Ramesh','Suresh',np.nan,'Mahesh'],
  'Age':[np.nan,22,23,25],
  'City':['hyd','blr','pune',np.nan]}
df2=pd.DataFrame(d)
df2.to_csv('data2.csv',index=False)
```

**KNNImputer**

- KNN : K-nearnest neighbours

- where K iis number

- instead of taking avg of all the value  data

    - first will find the neighbourd who are close with missing data

    - by using Euclidian distance calculation will identify the neighbour rows

    - suppose k=2 will identify the two closet rows 

    - finally will with mean

```python
salary=[50000,np.nan,70000,90000]
age=[25,28,30,40]
df=pd.DataFrame(zip(salary,age),
               columns=["Salary","Age"])
df
```

- sklearn 

    - impute

        - KNNimputer



```python
from sklearn.impute import KNNImputer
imputer=KNNImputer(n_neighbors=2)
pd.DataFrame(imputer.fit_transform(df))
```
